```python
#   https://youtu.be/haFgfkKqKMY?si=G5VJF9Ymf1w43ZyV 실습

import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://hpxl0518.pythonanywhere.com/static"
res = requests.get(url)
html = res.text
```

```python
print(res.status_code)
print(res.headers)
print(res.text)
```

```python
soup = BeautifulSoup(html, 'html.parser')
soup
```

```python
items = soup.select('.card')
items
```

```python
for item in items :
    title = item.select_one('.card-title').text
    print(title)
```
