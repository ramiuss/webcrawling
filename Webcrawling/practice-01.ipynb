```python
from bs4 import BeautifulSoup
html_txt = "<p class = 'weather' id='tw'>오늘의 날씨</p> <h1> 한 째 소나기가 내리겠습니다.</h1>"

soup = BeautifulSoup(html_txt, 'html.parser')
```

```python
tag = soup.find("p")

print(tag)
```

```python
print(tag.name)
```

```python
print(tag.attrs)
```

```python
print(tag.attrs['class'])
```

```python
print(tag.attrs['id'])
```

```python
print(tag.text)
```
