```python
import requests
from bs4 import BeautifulSoup

res = requests.get("https://product.kyobobook.co.kr/detail/S000217048850")
res.encoding = res.apparent_encoding  # 자동 추정된 인코딩으로 설정
html = res.text                       # 이제 문자열로 가져옴
book = BeautifulSoup(html, 'html.parser')
print(book)
```
