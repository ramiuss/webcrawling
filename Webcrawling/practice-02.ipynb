```python
from bs4 import BeautifulSoup

html_txt = """
<html>
<head><title>BS Pages</title></head>
<body>
<h1 class = "portal_cls">검색포텋</h1>
<p>
<a href = "http://www.daum.net">다음 바로가기</a>
<a href = "http://www.naver.com">네이버 바로가기</a>
</p>
<a href = "http://www.google.com" class="alink_cls">구글</a>
<a target = "_테스트_">테스트</a>
<p class = "footage_cls" id = "company">2021, ABC Company </p>
<p class = "footage_cls" id = "addr">Korea</p>

</body>
</html>
"""
soup = BeautifulSoup(html_txt, "html.parser")
```

```python
tag = soup.select_one("h1")
print(tag.text)
```

```python
tag_list = soup.select('h1')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup('body')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup.select('body p')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup.select('body p > a')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup.select('.footage_cls')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup.select('#company')
for tag in tag_list :
    print(tag.text)
```

```python
tag_list = soup.select('a[href]')
for tag in tag_list :
    print(tag.text, tag.attrs['href'])
```
